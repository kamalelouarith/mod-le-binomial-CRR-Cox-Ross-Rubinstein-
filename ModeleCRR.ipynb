```python
import numpy as np

def crr_european(S0, K, T, r, sigma, N, option='call'):
    # Tout ce bloc DOIT être indenté (décalé vers la droite)
    dt = T / N
    u = np.exp(sigma * np.sqrt(dt))
    d = 1 / u
    q = (np.exp(r * dt) - d) / (u - d)
    discount = np.exp(-r * dt)
    
    j = np.arange(N + 1)
    ST = S0 * u**j * d**(N - j)
    
    if option == 'call':
        payoff = np.maximum(ST - K, 0)
    elif option == 'put':
        payoff = np.maximum(K - ST, 0)
    else:
        raise ValueError("L'option doit être 'call' ou 'put'")
        
    for n in range(N - 1, -1, -1):
        payoff = discount * (q * payoff[1:] + (1 - q) * payoff[:-1])
        
    return payoff[0]
```

```python
S0, K, T, r, sigma = 100, 100, 1.0, 0.05, 0.20
for N in [10, 50, 100, 500, 1000]:
    price = crr_european(S0, K, T, r, sigma, N)
    print(f"N={N:4d}: C = {price:.4f}")
```
